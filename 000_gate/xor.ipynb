# AND

```python
import numpy as np
def AND(x1, x2):
    x = np.array([x1, x2])   #입력값
    w = np.array([0.5, 0.5]) #가중치
    b = -0.7                 #편향
    tmp = np.sum(w*x) + b
    if tmp > 0:
        return 1
    else:
        return 0
```

# OR


```python
def OR(x1, x2):
    x = np.array([x1, x2])   #입력값
    w = np.array([0.5, 0.5]) #가중치
    b = -0.2                 #편향
    tmp = np.sum(w*x) + b
    if tmp > 0:
        return 1
    else:
        return 0
```

# NAND

```python
def NAND(x1, x2):
    x = np.array([x1, x2])     #입력값
    w = np.array([-0.5, -0.5]) #가중치
    b = 0.7                    #편향
    tmp = np.sum(w*x) + b
    if tmp > 0:
        return 1
    else:
        return 0
```

# XOR

```python
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y
```

OR 퍼셉트론 테스트

```python
print(XOR(0, 0))
print(XOR(1, 0))
print(XOR(0, 1))
print(XOR(1, 1))
```
